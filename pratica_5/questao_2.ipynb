# Questão 2

Resumo sobre transformadas de wavelets e aplicações.


## Resumo

As transformadas de wavelets surgem como uma ferramenta de análise tempo-frequência com resolução multiescala.
Diferentemente da transformada de Fourier, que usa senos e cossenos globalmente definidos, a análise por wavelets
emprega funções localizadas no tempo, o que permite representar sinais não estacionários com muito mais flexibilidade.

Na prática, a decomposição wavelet separa um sinal em aproximações e detalhes. As aproximações concentram o conteúdo
de baixa frequência, enquanto os detalhes capturam transições rápidas, bordas, impulsos e estruturas locais. Essa ideia
é especialmente útil quando o comportamento do sinal varia ao longo do tempo.

## Principais vantagens

- representação multirresolução de sinais e imagens;
- boa localização simultânea no tempo e na frequência;
- eficiência para sinais com transientes e descontinuidades;
- compressão esparsa em muitos cenários práticos;
- suporte natural a técnicas de limiarização e remoção de ruído.

## Aplicações comuns

- compressão de imagens e áudio;
- remoção de ruído em sinais biomédicos, sísmicos e industriais;
- detecção de falhas, bordas e transientes;
- análise multiescala em séries temporais;
- telecomunicações, biometria, sensoriamento e visão computacional.

## Comentário final

O principal valor das wavelets está em adaptar a análise à escala do fenômeno observado. Por isso, elas são mais adequadas
do que a Fourier clássica quando o sinal contém mudanças locais importantes ou quando a estrutura relevante aparece em escalas distintas.
